```python
from galpy.orbit import Orbit
from galpy.potential import MWPotential2014
from potential import MW
from astropy.units import Unit
import numpy as np
```

```python
o = Orbit([1.0, 0.0, 1.0, 0.0, 0.83, 0.0], ro=8.34, vo=240)
```

```python
tt = np.linspace(0, 0.2, 1000) * Unit('Gyr')
o.integrate(tt, MWPotential2014)
```

```python
print(f"zmax = {o.zmax():.2f}")
```

```python
import matplotlib.pyplot as plt

plt.plot(tt.value, o.z(tt))
```
